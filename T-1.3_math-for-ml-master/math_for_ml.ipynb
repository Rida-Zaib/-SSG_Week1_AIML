## 1. Statistics
Given a list of exam scores, find the mean, variance, and standard deviation.

```python
import numpy as np

scores = [72, 85, 90, 65, 78, 88, 95, 60, 82, 75]

mean = np.mean(scores)
variance = np.var(scores)
std = np.std(scores)

print("mean:", mean)
print("variance:", variance)
print("std dev:", std)
```

## 2. Probability
Simulate flipping a fair coin 10,000 times and estimate the probability of heads. Also compute the exact probability of getting exactly 5 heads in 10 flips (binomial).

```python
import random
from math import comb

flips = [random.choice(["H", "T"]) for _ in range(10000)]
heads_ratio = flips.count("H") / len(flips)
print("simulated probability of heads:", heads_ratio)

n, k, p = 10, 5, 0.5
exact_prob = comb(n, k) * (p**k) * ((1-p)**(n-k))
print("exact probability of 5 heads in 10 flips:", exact_prob)
```

## 3. Linear Algebra
Given two matrices, compute their product, and find the determinant and inverse of one of them. This kind of operation shows up constantly in ML, for example in linear regression's normal equation.

```python
import numpy as np

A = np.array([[2, 0], [1, 3]])
B = np.array([[1, 2], [3, 4]])

product = A @ B
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)

print("A @ B =\n", product)
print("det(A) =", det_A)
print("inverse of A =\n", inv_A)
```

## 4. Calculus
Find the derivative of the function f(x) = 3x^2 + 2x - 5, and use it to find the slope at x = 4. This is the same idea used when computing gradients during model training.

```python
import sympy as sp

x = sp.symbols("x")
f = 3*x**2 + 2*x - 5
f_prime = sp.diff(f, x)

slope_at_4 = f_prime.subs(x, 4)

print("f(x) =", f)
print("f'(x) =", f_prime)
print("slope at x=4:", slope_at_4)
```

## Notes
- Statistics: variance and std tell us how spread out the scores are around the mean.
- Probability: the simulated result should land close to the exact 0.5 as the number of flips grows, this is the law of large numbers.
- Linear algebra: matrix multiplication and inverses are the backbone of how linear models solve for their weights.
- Calculus: the derivative gives the slope of the loss function, which is exactly what gradient descent follows during training.
